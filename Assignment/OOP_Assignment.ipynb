```python
# Calculator Class

class Calculator:

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def multiply(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "Error: Division by zero"
        return self.a / self.b


# Taking user input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Creating object
calc = Calculator(num1, num2)

# Performing operations
print("Addition:", calc.add())
print("Subtraction:", calc.subtract())
print("Multiplication:", calc.multiply())
print("Division:", calc.divide())
```

4. Write a Python program to create a class that represents a shape. Include methods to calculate its area and perimeter. Implement subclasses for different shapes like circle, triangle, and square.

```python
import math

# Base Class
class Shape:
    def area(self):
        pass

    def perimeter(self):
        pass


# Circle Class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


# Square Class
class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

    def perimeter(self):
        return 4 * self.side


# Triangle Class
class Triangle(Shape):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def area(self):
        # Heron's Formula
        s = (self.a + self.b + self.c) / 2
        return math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))

    def perimeter(self):
        return self.a + self.b + self.c


# Example usage
c = Circle(5)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

s = Square(4)
print("\nSquare Area:", s.area())
print("Square Perimeter:", s.perimeter())

t = Triangle(3, 4, 5)
print("\nTriangle Area:", t.area())
print("Triangle Perimeter:", t.perimeter())
```

8. Write a Python program to create a class representing a shopping cart. Include methods for adding and removing items, and calculating the total price.

```python
# Shopping Cart Class

class ShoppingCart:

    def __init__(self):
        self.items = {}   # Dictionary to store items {item_name: price}

    def add_item(self, name, price):
        self.items[name] = price
        print(f"{name} added to cart.")

    def remove_item(self, name):
        if name in self.items:
            del self.items[name]
            print(f"{name} removed from cart.")
        else:
            print("Item not found in cart.")

    def show_items(self):
        if not self.items:
            print("Cart is empty.")
        else:
            print("Items in cart:")
            for name, price in self.items.items():
                print(f"{name} - ₹{price}")

    def calculate_total(self):
        total = sum(self.items.values())
        return total


# Example usage
cart = ShoppingCart()

cart.add_item("Laptop", 50000)
cart.add_item("Mouse", 500)
cart.add_item("Keyboard", 1500)

cart.show_items()

print("\nTotal Price: ₹", cart.calculate_total())

cart.remove_item("Mouse")
cart.show_items()

print("\nUpdated Total: ₹", cart.calculate_total())
```

13. TechWorld, a technology training center, wants to allocate courses for instructors. An instructor is identified by name, technology skills, experience and average feedback. An instructor is allocated a course, if he/she satisfies the below two conditions:

	eligibility criteria:
	if experience is more than 3 years, average feedback should be 4.5 or more
	if experience is 3 years or less, average feedback should be 4 or more
	he/she should posses the technology skill for the course
	Identify the class name and attributes to represent instructors. Write a Python program to implement the class chosen with its attributes and methods.

	Note:

	Consider all instance variables to be private and methods to be public.
	An instructor may have multiple technology skills, so consider instance variable, technology_skill to be a list.
	check_eligibility(): Return true if eligibility criteria is satisfied by the instructor. Else, return false
	allocate_course(technology): Return true if the course which requires the given technology can be allocated to the instructor. Else, return false.
	Represent a few objects of the class, initialize instance variables using setter methods, invoke appropriate methods and test your program.


```python
class Instructor:

    def __init__(self):
        self.__name = None
        self.__technology_skills = []
        self.__experience = 0
        self.__average_feedback = 0.0

    # Setter methods
    def set_name(self, name):
        self.__name = name

    def set_technology_skills(self, skills):
        self.__technology_skills = skills

    def set_experience(self, exp):
        self.__experience = exp

    def set_average_feedback(self, feedback):
        self.__average_feedback = feedback

    # Getter methods
    def get_name(self):
        return self.__name

    def get_technology_skills(self):
        return self.__technology_skills

    def get_experience(self):
        return self.__experience

    def get_average_feedback(self):
        return self.__average_feedback

    # Check eligibility
    def check_eligibility(self):
        if self.__experience > 3:
            return self.__average_feedback >= 4.5
        else:
            return self.__average_feedback >= 4

    # Allocate course
    def allocate_course(self, technology):
        if self.check_eligibility() and technology in self.__technology_skills:
            return True
        return False


# ----------- Testing the class ------------

# Instructor 1
i1 = Instructor()
i1.set_name("Gauri")
i1.set_technology_skills(["Python", "Java", "SQL"])
i1.set_experience(5)
i1.set_average_feedback(4.6)

print("Instructor:", i1.get_name())
print("Eligible:", i1.check_eligibility())
print("Can teach Python:", i1.allocate_course("Python"))
print("Can teach C++:", i1.allocate_course("C++"))

print("\n---------------------\n")

# Instructor 2
i2 = Instructor()
i2.set_name("Rahul")
i2.set_technology_skills(["C", "C++"])
i2.set_experience(2)
i2.set_average_feedback(4.1)

print("Instructor:", i2.get_name())
print("Eligible:", i2.check_eligibility())
print("Can teach C++:", i2.allocate_course("C++"))
print("Can teach Java:", i2.allocate_course("Java"))
```

14. Write a program that uses datetime module within a class. Enter manufacturing date and expiry date of the product. The program must display the years, months and days that are left for expiry.

```python
from datetime import datetime

class Product:

    def __init__(self, mfg_date, exp_date):
        # Convert string to datetime object
        self.__mfg_date = datetime.strptime(mfg_date, "%d-%m-%Y")
        self.__exp_date = datetime.strptime(exp_date, "%d-%m-%Y")

    def time_left(self):
        today = datetime.today()

        if today > self.__exp_date:
            print("Product has already expired.")
        else:
            diff = self.__exp_date - today

            # Convert days into years, months, days (approx)
            days = diff.days
            years = days // 365
            months = (days % 365) // 30
            remaining_days = (days % 365) % 30

            print("Time left for expiry:")
            print(f"{years} years, {months} months, {remaining_days} days")


# ----------- Testing ------------

mfg = input("Enter Manufacturing Date (dd-mm-yyyy): ")
exp = input("Enter Expiry Date (dd-mm-yyyy): ")

p = Product(mfg, exp)
p.time_left()
```

	Marks is between 0 and 100 (both inclusive)
	A student qualifies for admission, if

```python
class Student:

    def __init__(self):
        self.__student_id = None
        self.__age = 0
        self.__marks = 0

    # Setter methods
    def set_student_id(self, sid):
        self.__student_id = sid

    def set_age(self, age):
        self.__age = age

    def set_marks(self, marks):
        self.__marks = marks

    # Getter methods
    def get_student_id(self):
        return self.__student_id

    def get_age(self):
        return self.__age

    def get_marks(self):
        return self.__marks

    # Validate data
    def validate_data(self):
        if self.__age > 20 and 0 <= self.__marks <= 100:
            return True
        return False

    # Check qualification
    def check_qualification(self):
        if self.validate_data() and self.__marks >= 65:
            return True
        return False


# ----------- Testing ------------

# Student 1
s1 = Student()
s1.set_student_id(101)
s1.set_age(22)
s1.set_marks(75)

print("Student ID:", s1.get_student_id())
print("Valid Data:", s1.validate_data())
print("Eligible for Admission:", s1.check_qualification())

print("\n----------------------\n")

# Student 2
s2 = Student()
s2.set_student_id(102)
s2.set_age(19)
s2.set_marks(80)

print("Student ID:", s2.get_student_id())
print("Valid Data:", s2.validate_data())
print("Eligible for Admission:", s2.check_qualification())

print("\n----------------------\n")

# Student 3
s3 = Student()
s3.set_student_id(103)
s3.set_age(25)
s3.set_marks(60)

print("Student ID:", s3.get_student_id())
print("Valid Data:", s3.validate_data())
print("Eligible for Admission:", s3.check_qualification())
```

16. Ice-Cream Scoops and Bowl shop

	Create a class Scoop which has one public property flavor and one private proptery price. Take flavor values during object creation.

	Create a class Bowl with private prperty scoop_list which will have list of scoopd object.

	Create a method add_scoops in Bowl class which will add any no of Scoop objects given as parameter and store it in scoops_list.

	Make getter and setter method for price property.

	Make a method display to display flavour and price of each Scoop in scoop_list and print total price of the bowl by adding all flavour scoops prices.

	Make a method sold in both Scoop class and Bowl class to print no of quantity sold.

```python
# Scoop Class
class Scoop:

    __count = 0   # class variable to track scoops sold

    def __init__(self, flavor):
        self.flavor = flavor      # public
        self.__price = 0          # private
        Scoop.__count += 1

    # Setter for price
    def set_price(self, price):
        self.__price = price

    # Getter for price
    def get_price(self):
        return self.__price

    # Sold method
    @classmethod
    def sold(cls):
        print("Total scoops sold:", cls.__count)


# Bowl Class
class Bowl:

    __count = 0   # class variable to track bowls sold

    def __init__(self):
        self.__scoop_list = []   # private list
        Bowl.__count += 1

    # Add multiple scoops
    def add_scoops(self, *scoops):
        for scoop in scoops:
            self.__scoop_list.append(scoop)

    # Display method
    def display(self):
        total = 0
        print("Scoops in Bowl:")

        for scoop in self.__scoop_list:
            print(f"Flavor: {scoop.flavor}, Price: ₹{scoop.get_price()}")
            total += scoop.get_price()

        print("Total Price of Bowl: ₹", total)

    # Sold method
    @classmethod
    def sold(cls):
        print("Total bowls sold:", cls.__count)


# ----------- Testing ------------

# Create Scoop objects
s1 = Scoop("Vanilla")
s1.set_price(50)

s2 = Scoop("Chocolate")
s2.set_price(60)

s3 = Scoop("Strawberry")
s3.set_price(55)

# Create Bowl
b1 = Bowl()

# Add scoops
b1.add_scoops(s1, s2, s3)

# Display details
b1.display()

print("\n--- Sales Info ---")
Scoop.sold()
Bowl.sold()
```


17. Create a Bus child class that inherits from the Vehicle class. The default fare charge of any vehicle is seating capacity * 100. If Vehicle is Bus instance, we need to add an extra 10% on full fare as a maintenance charge. So total fare for bus instance will become the final amount = total fare + 10% of the total fare.

Note: The bus seating capacity is 50. so the final fare amount should be 5500. You need to override the fare() method of a Vehicle class in Bus class.


```python
# Parent Class
class Vehicle:

    def __init__(self, capacity):
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100


# Child Class
class Bus(Vehicle):

    def __init__(self, capacity):
        super().__init__(capacity)

    # Overriding fare method
    def fare(self):
        base_fare = super().fare()
        maintenance = base_fare * 0.10
        return base_fare + maintenance


# ----------- Testing ------------

bus = Bus(50)

print("Seating Capacity:", bus.capacity)
print("Total Fare:", bus.fare())
```

	Sample Employee Data:
	"ADAMS", "E7876", 50000, "ACCOUNTING"
	"JONES", "E7499", 45000, "RESEARCH"
	"MARTIN", "E7900", 50000, "SALES"
	"SMITH", "E7698", 55000, "OPERATIONS"

	Use 'assign_department' method to change the department of an employee.
	Use 'print_employee_details' method to print the details of an employee.
	Use 'calculate_emp_salary' method takes two arguments: salary and hours_worked, which is the number of hours worked by the employee. If the number of hours worked is more than 50, the method computes overtime and adds it to the salary. Overtime is calculated as following formula:
	overtime = hours_worked - 50
	Overtime amount = (overtime * (salary / 50))


```python
class Employee:

    def __init__(self, emp_name, emp_id, emp_salary, emp_department):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.emp_salary = emp_salary
        self.emp_department = emp_department

    # Method to assign/change department
    def assign_department(self, new_department):
        self.emp_department = new_department

    # Method to calculate salary with overtime
    def calculate_emp_salary(self, hours_worked):
        if hours_worked > 50:
            overtime = hours_worked - 50
            overtime_amount = overtime * (self.emp_salary / 50)
            total_salary = self.emp_salary + overtime_amount
        else:
            total_salary = self.emp_salary

        return total_salary

    # Method to print employee details
    def print_employee_details(self):
        print("Employee ID:", self.emp_id)
        print("Name:", self.emp_name)
        print("Salary:", self.emp_salary)
        print("Department:", self.emp_department)
        print("-" * 30)


# ----------- Testing ------------

# Creating Employee objects
e1 = Employee("ADAMS", "E7876", 50000, "ACCOUNTING")
e2 = Employee("JONES", "E7499", 45000, "RESEARCH")
e3 = Employee("MARTIN", "E7900", 50000, "SALES")
e4 = Employee("SMITH", "E7698", 55000, "OPERATIONS")

# Print details
e1.print_employee_details()
e2.print_employee_details()
e3.print_employee_details()
e4.print_employee_details()

# Change department
e3.assign_department("MARKETING")

print("\nAfter changing department of MARTIN:\n")
e3.print_employee_details()

# Calculate salary with overtime
print("Salary with overtime:")
print("ADAMS:", e1.calculate_emp_salary(55))
print("JONES:", e2.calculate_emp_salary(45))
```

19. Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order.

	Perform the following tasks now:

	Now add items to the menu.
	Make table reservations.
	Take customer orders.
	Print the menu.
	Print table reservations.
	Print customer orders.
	Note: Use dictionaries and lists to store the data.

```python
class Restaurant:

    def __init__(self):
        self.menu_items = {}          # {item_name: price}
        self.book_table = []          # list of table numbers
        self.customer_orders = {}     # {customer_name: [items]}

    # Add items to menu
    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    # Book tables
    def book_tables(self, table_number):
        self.book_table.append(table_number)

    # Take customer orders
    def customer_order(self, customer_name, items):
        self.customer_orders[customer_name] = items

    # Print menu
    def print_menu(self):
        print("\n--- Menu ---")
        for item, price in self.menu_items.items():
            print(f"{item} : ₹{price}")

    # Print table reservations
    def print_table_reservations(self):
        print("\n--- Table Reservations ---")
        for table in self.book_table:
            print(f"Table {table} booked")

    # Print customer orders
    def print_customer_orders(self):
        print("\n--- Customer Orders ---")
        for customer, items in self.customer_orders.items():
            print(f"{customer} ordered: {', '.join(items)}")


# ----------- Testing ------------

r = Restaurant()

# Add items to menu
r.add_item_to_menu("Pizza", 250)
r.add_item_to_menu("Burger", 120)
r.add_item_to_menu("Pasta", 200)

# Book tables
r.book_tables(1)
r.book_tables(3)

# Take orders
r.customer_order("Gauri", ["Pizza", "Pasta"])
r.customer_order("Rahul", ["Burger"])

# Display data
r.print_menu()
r.print_table_reservations()
r.print_customer_orders()
```
